# Funciones

```python
from pathlib import Path
from collections import defaultdict
import numpy as np
from PIL import Image
from concurrent.futures import ProcessPoolExecutor
from tqdm.auto import tqdm


def contar_imagenes(rutas):
    total = 0
    for ruta in rutas:
        total_items = len(list(ruta.iterdir()))
        total = total + total_items   
        
    print(f"Total items: {total}")


def get_size(img_path):
    try:
        with Image.open(img_path) as img:
            return img.size
    except:
        return None


def ver_resoluciones(rutas):

    files = []
    for ruta in rutas:
        files.extend(ruta.iterdir())

    resolutions = set()

    with ProcessPoolExecutor() as executor:
        for size in tqdm(
            executor.map(get_size, files, chunksize=100),
            total=len(files),
            desc="Reading images"
        ):
            if size is not None:
                resolutions.add(size)

    print(f"\nFound {len(resolutions)} unique resolutions:")
    for w, h in sorted(resolutions):
        print(f"{w}x{h}")
```

```python
images_dir = Path("/kaggle/input/datasets/mrpeponee/rings-dataset/RINGS algorithm dataset/TRAIN/TRAIN/IMAGES")
# Realizar estritificacion por paciente y por categoria
image_files = list(images_dir.glob("*.png"))


values_by_pos = defaultdict(list)

for img_path in image_files:
    parts = img_path.stem.split("_")
    

    for pos, value in enumerate(parts):
        values_by_pos[pos].append(value)

# Contar únicos con numpy
for pos in sorted(values_by_pos):
    uniques = np.unique(values_by_pos[pos])

    print(f"\nPosición {pos}")
    print(f"Valores únicos: {len(uniques)}")
    print(uniques[:])  # primeros 20 para inspección
```

```python
import os
from collections import Counter
from tqdm.auto import tqdm

folders = [
    "/kaggle/input/datasets/whats2000/breast-cancer-semantic-segmentation-bcss/BCSS/train",
    "/kaggle/input/datasets/whats2000/breast-cancer-semantic-segmentation-bcss/BCSS/val",
    "/kaggle/input/datasets/whats2000/breast-cancer-semantic-segmentation-bcss/BCSS/test",
]

counter = Counter()

all_files = []

for folder in folders:
    for fname in os.listdir(folder):
        if fname.lower().endswith(".png"):
            all_files.append(fname)

print(f"Found {len(all_files):,} images")

for fname in tqdm(all_files, desc="Counting third values"):

    prefix = fname.split("_")[0]
    parts = prefix.split("-")

    if len(parts) >= 3:
        third_value = parts[2]  # A0SK
        counter[third_value] += 1

print(f"\nUnique third values: {len(counter)}\n")

print("Third value -> Number of images")
print("-" * 40)

for value, count in sorted(counter.items()):
    print(f"{value}: {count}")

print("\nSummary")
print("-" * 40)
print(f"Minimum images : {min(counter.values())}")
print(f"Maximum images : {max(counter.values())}")
print(f"Average images : {sum(counter.values()) / len(counter):.2f}")
```
